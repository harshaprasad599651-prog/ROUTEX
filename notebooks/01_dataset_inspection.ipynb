## 9. Missing values

A **missing value** (shown as `NaN`) means no information was recorded for that cell.
Most models cannot handle missing values, so we need to know:

- how many values are missing in each column, and
- what **percentage** of the column that represents.

A column that is 95% empty is very different from one with only a few gaps.
Here we only *measure* missing values. Deciding what to do about them is part of
preprocessing.

```python
missing_count = df.isna().sum()
missing_percent = (missing_count / len(df) * 100).round(2)

missing_table = (
    pd.DataFrame({
        "missing_count": missing_count,
        "missing_percent": missing_percent,
    })
    .sort_values("missing_percent", ascending=False)
)

columns_with_missing = missing_table[missing_table["missing_count"] > 0]
print(f"Columns with at least one missing value: {len(columns_with_missing)} of {df.shape[1]}")
print(f"Total missing cells: {missing_count.sum():,}")
columns_with_missing
```

## 10. Duplicate rows

A **duplicate row** is a row that is exactly the same as another row in every column.
Duplicates can give some vehicles extra weight during training and can make test
results look better than they really are (the same record could appear in both the
training and the test set).

We check two things:

1. Rows that are identical in **every** column.
2. Rows that are identical in every column **except `id`**. Each record has its own
   `id` number, so two records could describe the same vehicle even though their
   `id` values differ.

```python
exact_duplicates = df.duplicated().sum()
print(f"Exact duplicate rows (all columns): {exact_duplicates:,}")

if "id" in df.columns:
    print(f"Is 'id' unique for every row?    {df['id'].is_unique}")
    duplicates_ignoring_id = df.drop(columns="id").duplicated().sum()
    print(f"Duplicate rows ignoring 'id':      {duplicates_ignoring_id:,}")
```

## 12. Number of unique values per column

The number of **unique values** tells us what kind of information a column holds:

- **1 unique value**: the column is the same for every vehicle, so it cannot help a model.
- **A few unique values**: probably a category (e.g. `drive`, `fuelType`).
- **Almost as many unique values as rows**: probably an identifier (e.g. `id`) or a
  very detailed text field, which is usually not useful for prediction.

By default, `nunique()` does **not** count missing values. So we show two numbers:
`unique_values` (missing values ignored) and `unique_incl_missing` (missing counted as
one extra value). A column like `tCharger` has only one real value (`T` = turbocharged),
but the empty cells also carry meaning (not turbocharged), so it is **not** truly constant.

```python
unique_table = pd.DataFrame({
    "unique_values": df.nunique(),
    "unique_incl_missing": df.nunique(dropna=False),
    "data_type": df.dtypes.astype(str),
}).sort_values("unique_values", ascending=False)

unique_table
```

```python
# Truly constant: the same value (or always missing) in every row
constant_columns = unique_table[unique_table["unique_incl_missing"] <= 1].index.tolist()

# One real value plus missing cells: the "missing" part may carry meaning
one_value_plus_missing = unique_table[
    (unique_table["unique_values"] == 1) & (unique_table["unique_incl_missing"] == 2)
].index.tolist()

print(f"Truly constant columns:                  {constant_columns}")
print(f"Columns with one value + missing cells:  {one_value_plus_missing}")
for column in one_value_plus_missing:
    print(f"  {column}: values = {df[column].unique().tolist()}")
```

## 15. Checking for suspicious or invalid target values

We look for values that might be errors or that do not fit our problem. We **only count
and display** them here. Nothing is deleted.

| Check | Why it matters |
|-------|----------------|
| Zero or negative MPG | A real vehicle cannot have 0 or negative fuel economy, so this would be a data error or a placeholder. |
| Non-whole numbers | `comb08` is normally stored as a rounded whole number, so decimals would be unexpected. |
| Statistical outliers (IQR rule) | Values that are far away from most other values. |
| Electric vehicles | For electric cars, the "MPG" figure is actually **MPGe** (miles per gallon *equivalent*), which is calculated differently and is usually much higher than any petrol car. |

### 15.1 Zero, negative and non-whole values

```python
non_positive = (target <= 0).sum()
non_whole = (target.dropna() % 1 != 0).sum()

print(f"Values <= 0:                {non_positive}")
print(f"Values that are not whole:  {non_whole}")
```

### 15.2 Outliers using the IQR rule

The **interquartile range (IQR)** is the distance between the 25th percentile (Q1) and
the 75th percentile (Q3). It covers the middle 50% of the data. A common rule of thumb
says a value is an *outlier* if it is:

- below `Q1 - 1.5 × IQR`, or
- above `Q3 + 1.5 × IQR`.

An outlier is **not automatically wrong**. It may be a genuinely unusual vehicle.
We just want to know how many there are and what they look like.

```python
q1 = target.quantile(0.25)
q3 = target.quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

low_outliers = target < lower_bound
high_outliers = target > upper_bound

print(f"Q1 = {q1}, Q3 = {q3}, IQR = {iqr}")
print(f"Lower bound = {lower_bound}, Upper bound = {upper_bound}")
print(f"Values below lower bound: {low_outliers.sum():,}")
print(f"Values above upper bound: {high_outliers.sum():,}")
```

Let us look at a few of the vehicles with the **highest** `comb08` values, together with
some descriptive columns, to understand what kind of vehicles they are.

```python
context_columns = [c for c in ["year", "make", "model", "fuelType", "atvType", "VClass", TARGET]
                   if c in df.columns]

df.loc[high_outliers, context_columns].sort_values(TARGET, ascending=False).head(10)
```

### 15.3 Target values by fuel type

Grouping `comb08` by `fuelType` shows whether some fuel types (especially electricity)
have very different values from the rest.

```python
if "fuelType" in df.columns:
    target_by_fuel = (
        df.groupby("fuelType")[TARGET]
        .agg(["count", "mean", "min", "max"])
        .round(1)
        .sort_values("mean", ascending=False)
    )
    display(target_by_fuel)
```

```python
if "fuelType" in df.columns:
    is_electric = df["fuelType"].astype(str).str.contains("Electricity", case=False, na=False)
    print(f"Vehicles whose fuel type mentions electricity: {is_electric.sum():,} "
          f"({is_electric.mean() * 100:.2f}% of rows)")
    print(f"Outliers above the upper bound that are electric: "
          f"{(high_outliers & is_electric).sum():,} of {high_outliers.sum():,}")
```

### 15.4 Columns closely related to the target

Some columns describe the **same fuel-economy measurement** as `comb08` in a different
form, for example city MPG (`city08`), highway MPG (`highway08`), annual fuel cost
(`fuelCost08`), fuel consumption in barrels (`barrels08`) or CO2 emissions
(`co2TailpipeGpm`).

If we used these as inputs, the model would effectively be given the answer. This is
called **data leakage**. Here we simply measure how strongly they are related to
`comb08` using the correlation coefficient (from -1 to +1). We will decide which
columns to exclude during feature selection.

```python
related_columns = [c for c in ["city08", "highway08", "UCity", "UHighway",
                               "fuelCost08", "barrels08", "co2TailpipeGpm", "youSaveSpend"]
                   if c in df.columns]

df[related_columns + [TARGET]].corr(numeric_only=True)[TARGET].drop(TARGET).round(3)
```
